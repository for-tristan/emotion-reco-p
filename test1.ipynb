```python
def count_images(path):
    total = 0

    if not os.path.exists(path):
        print("Path not found:", path)
        return 0

    classes = sorted([c for c in os.listdir(path) if os.path.isdir(os.path.join(path, c))])

    for c in classes:
        folder = os.path.join(path, c)

        num_images = len(os.listdir(folder))
        print(f"{c}: {num_images}")

        total += num_images

    print("Number of classes:", len(classes))

    print("-" * 30)
    return total


train_total = count_images("C:\\Cpyprojects\\test")
test_total = count_images("C:\\Cpyprojects\\test")

print("Train Total:", train_total)
print("Test Total:", test_total)
print("Total:", train_total + test_total)
```
